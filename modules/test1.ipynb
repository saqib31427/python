```python
#exception handling:it is a way to handle the error that we predict using try and except block
#normal flow of the program is called try block and the code that handle the error is called except block
#exception flow of the program is called except block and the code that handle the error is called try block
#why use exception handling: it allow us to focus on the main task of our program instead of worrying about the error that may occur
#try block: it is the block of code that we want to execute and we predict that it may raise an error
#except block: it is the block of code that we want to execute if an error occurs
#EXTRA code mindset: it is a way to handle the error that we predict using try and except block 
#exception handling mindset: it is a way to handle the error that we predict using try and except block
```

```python
try:
   with open("C:\\Users\\HP\\python\\modules\\minot1.txt",'r') as data :
    for i in data :
     if ":" in i :
      try:
          speaker, ls= i.split(":")
          print(speaker,"said",ls)
      except:
       pass  
except:
    print("the file is missing")
```
